```python
import cv2
import os
import matplotlib.pyplot as plt
```

```python
input_folder = r"C:\Users\TECH STORE\OneDrive\Desktop\IP_Project\frames\match1"
output_folder = r"C:\Users\TECH STORE\OneDrive\Desktop\IP_Project\grayscale_frames\match1"

# create folder if it does not exist
os.makedirs(output_folder, exist_ok=True)
```

```python
frame_files = sorted(os.listdir(input_folder))

print("Total frames found:", len(frame_files))
```

```python
for file in frame_files:
    
    if file.endswith(".jpg") or file.endswith(".png"):
        
        image_path = os.path.join(input_folder, file)
        image = cv2.imread(image_path)
        
        gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
        
        output_path = os.path.join(output_folder, file)
        cv2.imwrite(output_path, gray)

print("Grayscale conversion completed.")
```

```python
sample_color = cv2.imread(r"C:\Users\TECH STORE\OneDrive\Desktop\IP_Project\frames\match1\frame_0000.jpg")
sample_gray = cv2.imread(r"C:\Users\TECH STORE\OneDrive\Desktop\IP_Project\grayscale_frames\match1\frame_0000.jpg", 0)

plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(cv2.cvtColor(sample_color, cv2.COLOR_BGR2RGB))
plt.title("Original Frame")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(sample_gray, cmap="gray")
plt.title("Grayscale Frame")
plt.axis("off")
```
