```python
import cv2
import os
import matplotlib.pyplot as plt
```

```python
video_path = r"C:\Users\TECH STORE\OneDrive\Desktop\IP_Project\videos\match1.mp4"
output_folder = r"C:\Users\TECH STORE\OneDrive\Desktop\IP_Project\frames\match1"

os.makedirs(output_folder, exist_ok=True)
```

```python
cap = cv2.VideoCapture(video_path)

if not cap.isOpened():
    print("Error: Cannot open video file")
else:
    print("Video loaded successfully")
```

```python
video_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
video_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
fps = cap.get(cv2.CAP_PROP_FPS)

sample_rate = 3
frames_tested = total_frames / sample_rate

print("Test Dataset Details - Match 1")
print("Resolution:", video_width, "x", video_height)
print("Frame Rate:", fps, "FPS")
print("Total Frames in Video:", total_frames)
print("Frames Tested/Extracted:", frames_tested)
print("Environmental Conditions: Outdoor sports video, fast player motion, camera movement, changing lighting")
```

```python
frame_count = 0
saved_count = 0
sample_rate = 3

while True:
    
    ret, frame = cap.read()
    
    if not ret:
        break
    
    if frame_count % sample_rate == 0:
        
        filename = f"{output_folder}/frame_{saved_count:04d}.jpg"
        cv2.imwrite(filename, frame)
        
        saved_count += 1
    
    frame_count += 1

cap.release()

print("Frames extracted:", saved_count)
```
