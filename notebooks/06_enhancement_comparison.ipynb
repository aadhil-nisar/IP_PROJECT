```python
import cv2
import os
import matplotlib.pyplot as plt
```

```python
gray_folder = r"C:\Users\TECH STORE\OneDrive\Desktop\IP_Project\grayscale_frames\match1"
denoise_folder = r"C:\Users\TECH STORE\OneDrive\Desktop\IP_Project\denoised_frames\match1"
enhanced_folder = r"C:\Users\TECH STORE\OneDrive\Desktop\IP_Project\enhanced_frames\match1"
```

```python
frame_name = "frame_0000.jpg"

gray_path = os.path.join(gray_folder, frame_name)
denoise_path = os.path.join(denoise_folder, frame_name)
enhanced_path = os.path.join(enhanced_folder, frame_name)

gray_img = cv2.imread(gray_path, 0)
denoise_img = cv2.imread(denoise_path, 0)
enhanced_img = cv2.imread(enhanced_path, 0)
```

```python
plt.figure(figsize=(15,5))

plt.subplot(1,3,1)
plt.imshow(gray_img, cmap="gray")
plt.title("Grayscale Frame")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(denoise_img, cmap="gray")
plt.title("After Gaussian Denoising")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(enhanced_img, cmap="gray")
plt.title("After CLAHE Enhancement")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(15,5))

plt.subplot(1,3,1)
plt.hist(gray_img.ravel(), bins=256, range=(0,256))
plt.title("Grayscale Histogram")

plt.subplot(1,3,2)
plt.hist(denoise_img.ravel(), bins=256, range=(0,256))
plt.title("Denoised Histogram")

plt.subplot(1,3,3)
plt.hist(enhanced_img.ravel(), bins=256, range=(0,256))
plt.title("CLAHE Histogram")

plt.tight_layout()
plt.show()
```
